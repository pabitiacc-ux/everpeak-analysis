```python
#cargar archivos
traffic = pd.read_csv('/datasets/tomtom_traffic.csv')
eco = pd.read_csv('/datasets/oecd_city_economy.csv')
```

```python
#Excaminar la estructura de Eco
eco.info()
eco.head(3)
```

** En eco se necesita convertir los campos City GDP/capita,Unemployment % y Population a float, ya que actualmente en la estructura se encuentran como string

```python
# Estandarizas la columnas de traffic
traffic = traffic.rename(columns={
    'Country': 'country',
    'City': 'city',
    'UpdateTimeUTC': 'update_time_utc',
    'JamsDelay': 'jams_delay',
    'TrafficIndexLive': 'traffic_index_live',
    'JamsLengthInKms': 'jams_length_kms',
    'JamsCount': 'jams_count',
    'TrafficIndexWeekAgo': 'traffic_index_week_ago',
    'UpdateTimeUTCWeekAgo': 'update_time_utc_week_ago',
    'TravelTimeLivePer10KmsMins': 'travel_time_live_per_10kms_mins',
    'TravelTimeHistoricPer10KmsMins': 'travel_time_historic_per_10kms_mins',
    'MinsDelay': 'mins_delay'
})
traffic.columns
```

```python
# Estandarizar los nombres de las columnas de eco
eco = eco.rename(columns={
    'Year': 'year',
    'City': 'city',
    'Country': 'country',
    'City GDP/capita': 'city_gdp_capita',
    'Unemployment %': 'unemployment_pct',
    'PM2.5 (μg/m³)': 'pm25',
    'Population (M)': 'population_m'
})

eco.columns
```

```python
# Convertir las columnas de traffic a tipo fecha con pd.to_datetime()
traffic['update_time_utc'] = pd.to_datetime(traffic["update_time_utc"], errors='coerce')
traffic['update_time_utc_week_ago'] = pd.to_datetime(traffic["update_time_utc_week_ago"], errors='coerce')  #tu código aquí

# verificar el cambio
traffic.info()
```

```python
# Limpia separadores y convierte columnas numéricas en eco
eco['city_gdp_capita'] = eco['city_gdp_capita'].astype(str).str.replace('.', '').str.replace(',', '.').astype(float)
eco['unemployment_pct'] = eco['unemployment_pct'].astype(str).str.replace('%', '', regex=False).str.replace(',', '.', regex=False).astype(float)
eco['population_m'] = eco['population_m'].astype(str).str.replace('.', '').str.replace(',', '.').astype(float)

# Calcula la población total en unidades absolutas (Multiplica * 1000000)
eco['population'] = eco['population_m'] * 1000000

# verificar el cambio
eco.info()
eco.head(3)
```

```python
# Extraer el año de las fechas en update_time_utc
traffic['year'] = traffic['update_time_utc'].dt.year

# Verificar cambio
traffic.head(3)
```

```python
# Filtra los registros del año 2024
traffic_2024 = traffic[traffic['year'] == 2024].copy()
eco_2024 = eco[eco['year']==2024].copy()

# Revisar dataframes nuevos
display(traffic_2024.head())
display(eco_2024.head())
```

```python
#Paso 4: Calcular los  promedios de trafico por ciudad, país y año
traffic_city_year_2024 = traffic_2024.groupby(['city', 'country', 'year']).agg({
    'jams_delay': 'mean',
    'traffic_index_live': 'mean',
    'jams_length_kms': 'mean',
    'jams_count': 'mean',
    'mins_delay': 'mean',
    'travel_time_live_per_10kms_mins': 'mean',
    'travel_time_historic_per_10kms_mins': 'mean'
}).reset_index()

traffic_city_year_2024.head()
```

*** Momento de reflexion ***

```python
traffic_city_year_2024.sort_values(["jams_delay"], ascending=False)
```

Momento de reflexión: Al ordenar los datos de forma descendente, se observa que la ciudad con el mayor tiempo promedio de retraso por tráfico es Ciudad de México (mexico-city) con aproximadamente 2,833.06 minutos, superando por casi 700 minutos al segundo lugar (Tokio). Esta ciudad pertenece a Latinoamérica (México), destacando por encima de otras grandes urbes de Asia, Norteamérica y Europa presentes en el top 5

La ciudad con el mayor tiempo promedio de tráfico es  Ciudad de México

```python
# Paso 5: Seleccionar columnas clave de tráfico y economía
left_cols = ['city','country','year','jams_delay','traffic_index_live',
             'jams_length_kms','jams_count','mins_delay',
             'travel_time_live_per_10kms_mins','travel_time_historic_per_10kms_mins']

right_cols = ['city','year','city_gdp_capita','unemployment_pct','pm25','population']

# Usar .copy() para crear los dos nuevos datasets reducidos
traffic_2024_small = traffic_city_year_2024[left_cols].copy()
eco_2024_small = eco_2024[right_cols].copy()

# Unir datasets
merged = pd.merge(traffic_2024_small,eco_2024_small, on=['city','year'], how='inner')   # tu código aquí

# Mostrar las primeras 5 filas
merged.head()  # tu código aquí
```

```python
#Punto 6
# Crear boxplot para observar el comportamiento de los minutos de congestión JamsDelay
plt.figure(figsize=(8, 5))
sns.boxplot(y=merged['jams_delay'], showmeans=True, color='skyblue')

# obtener promedio para mostrarlo en título
mean_value = merged['jams_delay'].mean()
plt.title(f'Boxplot de JamsDelay (2024)\nPromedio: {mean_value:.2f}')
plt.ylabel('Jams Delay (minutos)')
plt.show()
```

```python
# Crear histograma para ver la distribución de la economía (city_gdp_capita)
plt.figure(figsize=(8, 5))
sns.histplot(merged['city_gdp_capita'], kde=True, bins=10, color='teal')
plt.title('Distribución del PIB per cápita en ciudades (2024)')
plt.xlabel('City GDP per capita (USD)')
plt.ylabel('Frecuencia')
plt.show()
```

```python
# Gráfico de barras comparativo
merged.set_index('city')[['jams_delay', 'city_gdp_capita']].plot(kind='bar', figsize=(12, 6), secondary_y='city_gdp_capita')
plt.title('Comparativa de Retraso por Tráfico y PIB per cápita por Ciudad (2024)')
plt.xlabel('Ciudad')
plt.xticks(rotation=90)
plt.show()
```

¿Las ciudades con mayor PIB per cápita también presentan más congestión?
No. Si bien algunas urbes con economías desarrolladas enfrentan problemas de tráfico, la riqueza económica por habitante no es el único factor determinante del volumen de embotellamientos.

¿O sucede lo contrario, o no existe una relación clara? No hay una correlación clara. Se identifican ciudades con alta congestión (como Ciudad de México) que contrastan con otras economías donde el tráfico está más mitigado. Esto sugiere que el problema de la movilidad urbana depende de factores estructurales específicos —como la inversión en transporte público, la planeación urbana y la infraestructura vial— más allá de la productividad económica individual de cada ciudad.

***** Resumen ejecutivo ****

- Contexto & objetivo:
Este análisis evalúa cómo la movilidad urbana (medida a través de la congestión vehicular y los tiempos de viaje) se relaciona con la productividad económica (PIB per cápita) en las principales
ciudades latinoamericanas y globales. Para ello, se utilizaron variables clave de tráfico de TomTom y datos socioeconómicos de la OECD, con el fin de identificar patrones que orienten futuras decisiones e 
inversiones en infraestructura de transporte.

- Cobertura de datos /  Metodología:
El estudio consolidó registros correspondientes al año 2024. Metodológicamente, el proceso abarcó la limpieza de datos, estandarización de columnas a formato snake_case, conversión de tipos de fecha y corrección de formatos numéricos (PIB, desempleo y población). Posteriormente, se calcularon los promedios anuales de tráfico por ciudad y se integraron ambas fuentes mediante una unión de tipo inner join, complementada con análisis visuales (boxplots, histogramas y gráficos comparativos) para detectar tendencias y valores atípicos.

- Hallazgos iniciales / Recomendaciones:
Los resultados evidencian que metrópolis como Ciudad de México registran los mayores tiempos de retraso por congestión. liderando el ranking regional. Sin embargo, no se identifica una correlación lineal estricta entre un mayor PIB per cápita y niveles más altos de tráfico, lo que demuestra que los problemas de movilidad obedecen a factores estructurales de planeación urbana. Se recomienda priorizar la inversión en transporte público eficiente en ciudades densamente congestionadas y profundizar en estudios de impacto vial específicos por región.
